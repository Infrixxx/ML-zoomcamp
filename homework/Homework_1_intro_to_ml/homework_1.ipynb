```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("car_fuel_efficiency_2026.csv")
df.head()
```

```python
#Getting total rows

total_rows = len (df)
print(total_rows)
```

```python
fuel_types=df['fuel_type'].nunique()
print(f"There are {fuel_types} fuel types")
```

```python
null_colum = (df.isnull().sum() > 0 ).sum()
print(f"There are {null_colum} columns with missing values")
```

```python
fuel_eff_max =df['fuel_efficiency_mpg'].max()
print(f"max fuel efficiency is {fuel_eff_max} .")
```

```python
# Get the initial median before making any changes
initial_hp_median = df['horsepower'].median()
print(f"Initial median horsepower: {initial_hp_median}")

# Find the most frequent value
most_frequent_hp = df['horsepower'].mode()[0]
print(f"Most frequent horsepower used to fill blanks: {most_frequent_hp}")

# Fill the missing values
df['horsepower'] = df['horsepower'].fillna(most_frequent_hp)

# Get the new median
new_hp_median = df['horsepower'].median()
print(f"New median horsepower: {new_hp_median}")

# Automatically answer the question
if initial_hp_median == new_hp_median:
    print("Answer: No")
elif new_hp_median > initial_hp_median:
    print("Answer: Yes, it increased")
else:
    print("Answer: Yes, it decreased")
```

```python
# 1. Select all the cars from Asia
df_asia = df[df['origin'] == 'Asia']

# 2 & 3. Select only columns vehicle_weight and model_year, and get the first 7 values
df_subset = df_asia[['vehicle_weight', 'model_year']].iloc[:7]

# 4. Get the underlying NumPy array X
X = df_subset.to_numpy()

# 5. Compute matrix-matrix multiplication between the transpose of X and X
XTX = X.T @ X 

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 7. Create array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Multiply the inverse of XTX with the transpose of X, then by y to get w
w = XTX_inv @ X.T @ y

# 9. Calculate the sum of all elements in w
sum_w = w.sum()
print(f"The sum of weights (w) is: {sum_w}")
```
